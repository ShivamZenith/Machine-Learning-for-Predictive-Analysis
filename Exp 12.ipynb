```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the data
df = pd.read_csv(r"C:\Users\VICTUS\OneDrive\Desktop\Machine Learning for Predective Analysis\IRIS.csv")

print("=== FIRST 10 ROWS ===")
print(df.head(10))

print("\n=== DATASET INFO ===")
df.info()

print("\n=== BASIC STATISTICS ===")
print(df.describe())

print("\n=== MISSING VALUES ===")
print(df.isnull().sum())

print("\n=== DATA TYPES ===")
print(df.dtypes)

print("\n=== SHAPE OF DATASET ===")
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")

print("\n=== UNIQUE VALUES IN EACH COLUMN ===")
for col in df.columns:
    print(f"{col}: {df[col].nunique()} unique values")

# Visualization 1: Correlation Heatmap
print("\n=== GENERATING VISUALIZATIONS ===")
plt.figure(figsize=(10, 8))
numeric_cols = df.select_dtypes(include=[np.number]).columns
correlation_matrix = df[numeric_cols].corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0)
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()

# Visualization 2: Distribution of numerical features
df[numeric_cols].hist(figsize=(12, 10), bins=20)
plt.suptitle('Distribution of Numerical Features')
plt.tight_layout()
plt.show()

# Visualization 3: Box plots for numerical features
plt.figure(figsize=(12, 8))
df[numeric_cols].boxplot()
plt.title('Box Plot of Numerical Features')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Visualization 4: Pairplot (if Species column exists)
if 'Species' in df.columns:
    sns.pairplot(df, hue='Species', diag_kind='hist')
    plt.suptitle('Pairplot by Species', y=1.02)
    plt.show()

# Visualization 5: Count plot for categorical columns
categorical_cols = df.select_dtypes(include=['object']).columns
for col in categorical_cols:
    plt.figure(figsize=(8, 6))
    df[col].value_counts().plot(kind='bar')
    plt.title(f'Distribution of {col}')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

print("\n=== OUTLIER DETECTION ===")
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    print(f"{col}: {len(outliers)} outliers")

print("\n=== MEMORY USAGE ===")
print(f"Memory usage: {df.memory_usage(deep=True).sum() / 1024:.2f} KB")

print("\n=== EDA COMPLETED SUCCESSFULLY ===")
```
