```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Use the correct file path for your user
df = pd.read_csv(r"C:\Users\VICTUS\OneDrive\Desktop\Machine Learning for Predective Analysis\IRIS.csv")
print("First 10 rows:")
print(df.head(10))
print("\nDataset info:")
df.info()

# Missing values Handling (fixed version)
for col in df.columns:
    if df[col].dtype in ['int64', 'float64']:
        df[col] = df[col].fillna(df[col].median())
    else:
        df[col] = df[col].fillna(df[col].mode()[0])
print("Missing values after filling:\n", df.isnull().sum())

# Handling Duplicate Values
duplicate_count = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicate_count}")
df = df.drop_duplicates()

print(f"Data shape after removing duplicates: {df.shape}")

# Handling Null values
print("Null values before dropping:\n", df.isnull().sum())
df = df.dropna()

print(f"Shape after dropping nulls: {df.shape}")
print("Null values after dropping:\n", df.isnull().sum())

# Handling outliers for SepalLengthCm
Q1 = df['SepalLengthCm'].quantile(0.25)
Q3 = df['SepalLengthCm'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
df = df[(df['SepalLengthCm'] >= lower_bound) & (df['SepalLengthCm'] <= upper_bound)]
print(f"Data shape after removing outliers: {df.shape}")

# Final cleaned data
print("Final cleaned data:")
print(df.head())

# Additional: Check if any outliers were actually removed
print(f"\nOriginal data shape: 150 rows")
print(f"Final data shape: {df.shape[0]} rows")
print(f"Rows removed: {150 - df.shape[0]}")
```
